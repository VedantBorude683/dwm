```python
import pandas as pd 
df=pd.read_csv("titanic.csv")
```

```python
df.head()
```

```python
df.isnull().sum()
```

```python
df.dtypes
```

```python
print("mean",df["age"].mean())
print("median",df["age"].median())
print("mode",df["age"].mode()[0])#mean>median>mode so skewed data
df["age"]=df["age"].fillna(df["age"].median())
```

```python
df.isnull().sum()
```

```python
cat_col=["sex","embarked","class","who","deck","embark_town","alive"]#no need to put adult_male snd alone in category as they already represnt bool
df[cat_col]=df[cat_col].astype("category")
```

```python
df.dtypes
```

```python
df.isnull().sum()
```

```python
#embarked contains 2 missing values so fill with mode(category->mode)
df["embarked"]=df["embarked"].fillna(df["embarked"].mode()[0])
```

```python
df.isnull().sum()
```

```python
df["deck"]=df["deck"].astype("object").fillna("unknown")#since it has huge missing values so we would rather assign the values as unknown  
```

```python
df.isnull().sum()
```

```python
df["embark_town"]=df["embark_town"].fillna(df["embark_town"].mode()[0])
```

```python
df.isnull().sum()
```

```python
df.head()
```

```python
cat_col=["sex","embarked","class","who","deck","embark_town","alive"]
df=pd.get_dummies(df,columns=cat_col)
df.head()
```

```python

# who to scale and not to
#1-pclass-no becuase ordinal(ranking 3 2 1)
#2-age-yes(float)
#3sibsp and parch yes beacuse they are normal integer varibale
#4-fare yes
from sklearn.preprocessing import StandardScaler
scal=StandardScaler()
num_col=["age","sibsp","parch","fare"]
df[num_col]=scal.fit_transform(df[num_col])
```
