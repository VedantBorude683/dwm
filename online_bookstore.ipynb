```python
transactions = [
    ['Python Programming', 'Data Structures and Algorithms',
     'Database Management Systems'],

    ['Python Programming', 'Data Structures and Algorithms',
     'Operating Systems'],

    ['Python Programming', 'Data Structures and Algorithms',
     'Database Management Systems', 'Operating Systems'],

    ['Python Programming', 'Database Management Systems',
     'Machine Learning'],

    ['Python Programming', 'Data Structures and Algorithms',
     'Database Management Systems']
]
```

```python
import pandas as pd
from mlxtend.preprocessing import TransactionEncoder

te = TransactionEncoder()

encoded_data = te.fit(transactions).transform(transactions)

df = pd.DataFrame(encoded_data, columns=te.columns_)
```

```python
display(df)
```

```python
from mlxtend.frequent_patterns import apriori

frequent_itemsets = apriori(
    df,
    min_support=0.40,
    use_colnames=True
)

frequent_itemsets
```

```python
from mlxtend.frequent_patterns import association_rules

rules = association_rules(
    frequent_itemsets,
    metric="confidence",
    min_threshold=0.60
)

rules
```
