```python
transactions = [
    ['Burger', 'French Fries', 'Soft Drink'],
    ['Burger', 'French Fries'],
    ['Sandwich', 'Soft Drink'],
    ['Burger', 'Sandwich', 'Soft Drink'],
    ['Burger', 'French Fries', 'Ice Cream'],
    ['Sandwich', 'French Fries', 'Soft Drink'],
    ['Burger', 'French Fries', 'Soft Drink'],
    ['Sandwich', 'Soft Drink', 'Ice Cream'],
    ['Burger', 'French Fries', 'Sandwich'],
    ['Burger', 'Soft Drink']
]
```

```python
import pandas as pd
from mlxtend.preprocessing import TransactionEncoder
from mlxtend.frequent_patterns import apriori, association_rules
```

```python
#perform hot encoding
te = TransactionEncoder()

encoded_data = te.fit(transactions).transform(transactions)

df = pd.DataFrame(encoded_data, columns=te.columns_)

df.index = [f'Order {i}' for i in range(1, len(df) + 1)]

display(df)
```

```python
frequent_itemsets = apriori(
    df,
    min_support=0.30,
    use_colnames=True
)

frequent_itemsets
```

```python
rules = association_rules(
    frequent_itemsets,
    metric="confidence",
    min_threshold=0.50
)

rules
```

```python
result = rules[
    ['antecedents', 'consequents',
     'support', 'confidence', 'lift']
]

display(result)
```
