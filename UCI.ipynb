```python
col=["sex","address","famsize","Pstatus","guardian","schoolsup","paid","activities","nursery","higher","internet","romantic"]
df[col]=df[col].astype("category")
```

```python
#note ->we don;t change ordinal attributes like medu,fedu beacuse they alredy contain meaningful order
#change only nominal attributes
df_encoded=pd.get_dummies(df,columns=["Mjob","Fjob","reason","guardian"],dtype=int)
df_encoded.head()
```

```python
df_preprocessed=df_encoded.copy()
```

```python
from sklearn.preprocessing import MinMaxScaler
numerical_cols=df_preprocessed.select_dtypes(
    include='number'
).columns
scaler=MinMaxScaler()
df_preprocessed[numerical_cols]=scaler.fit_transform(
    df_preprocessed[numerical_cols]
)
```
