```python
import pandas as pd
from mlxtend.preprocessing import TransactionEncoder
from mlxtend.frequent_patterns import apriori,association_rules
```

```python
transactions=[
    ["Python","DBMS","AI"],
     ["Python","DBMS"],
     ["DBMS","AI"],
     ["Python","DBMS","AI"],
]
```

```python
te=TransactionEncoder()
encoded_data=te.fit(transactions).transform(transactions)
df=pd.DataFrame(encoded_data,columns=te.columns_)
print(df)
```

```python
frequent_itemset=apriori(
    df,
    min_support=0.4,
    use_colnames=True
)
print(frequent_itemset)
```

```python
rules=association_rules(
    frequent_itemset,
    metric="confidence",
    min_threshold=0.6
)
print(
    rules[
        ["antecedents","consequents","support","confidence"]
        ]
)
```

```python
print(frequent_itemset)
```

```python
print("frequent itemsets")
frequent_itemset["support"]=frequent_itemset["support"]*100
print(frequent_itemset.to_string(index=False))

print("\nAssociation rules")
rules["support"]=rules["support"]*100
rules["confidence"]=rules["confidence"]*100
print(
    rules[["antecedents","consequents","support","confidence"]]
    .round(2)
    .to_string(index=False)
)
```
