```python
import pandas as pd
df=pd.read_csv("train_loan_prediction.csv")
df.head()
```

```python
print("credithistory")
print("mean: ",df["Credit_History"].mean())
print("median: ",df["Credit_History"].median())
print("mode: ",df["Credit_History"].mode()[0])

print("loanamount")
print("mean: ",df["LoanAmount"].mean())
print("median: ",df["LoanAmount"].median())
print("mode: ",df["LoanAmount"].mode()[0])

print("loan_amount term")
print("mean: ",df["Loan_Amount_Term"].mean())
print("median: ",df["Loan_Amount_Term"].median())
print("mode: ",df["Loan_Amount_Term"].mode()[0])
```

```python
df.tail()
```

```python
df["LoanAmount"]=df["LoanAmount"].fillna(df["LoanAmount"].median())
```

```python
df["Credit_History"]=df["Credit_History"].fillna(df["Credit_History"].mode()[0])
df["Credit_History"]=df["Credit_History"].astype("category")
```

```python
df["Loan_Amount_Term"]=df["Loan_Amount_Term"].fillna(df["Loan_Amount_Term"].mode()[0])
df["Loan_Amount_Term"]=df["Loan_Amount_Term"].astype("category")
```

```python
df.isnull().sum()
```

```python
df.dtypes
```

```python
categorical_cols=[
    "Gender","Married","Dependents","Education","Self_Employed"
]
for col in categorical_cols:
    if df[col].isnull().sum()>0:
        df[col]=df[col].fillna(df[col].mode()[0])   
    df[col]=df[col].astype("category")
```

```python
df.isnull().sum()
```

```python
x=df.drop(columns=["Loan_ID","Loan_Status"])
y=df["Loan_Status"].map({"Y":1,"N":0})
print(x)
print(y)
```

```python
x=pd.get_dummies(x,drop_first=True,dtype=int)
x.head()
print(x.shape)
print(y.value_counts())
```

```python
from sklearn.model_selection import train_test_split
x_train,x_test,y_train,y_test=train_test_split(
    x,y,
    test_size=0.2,
    random_state=42,
    stratify=y
)
```

```python
x=pd.get_dummies(x,drop_first=True,dtype=int)
x.head()
```

```python
y=y.map({"Y":1,"N":0})
y.value_counts()
```

```python
from sklearn.tree import DecisionTreeClassifier
dt_model = DecisionTreeClassifier(random_state=42)

dt_model.fit(x_train, y_train)
```

```python
y_pred_dt = dt_model.predict(x_test)

print(y_pred_dt)
```

```python
from sklearn.metrics import accuracy_score, classification_report, confusion_matrix

print("Decision Tree Accuracy:", accuracy_score(y_test, y_pred_dt))

print("\nClassification Report:")
print(classification_report(y_test, y_pred_dt))

print("\nConfusion Matrix:")
print(confusion_matrix(y_test, y_pred_dt))
```

```python
from sklearn.naive_bayes import GaussianNB
nb_model = GaussianNB()

nb_model.fit(x_train, y_train)
```

```python
y_pred_nb = nb_model.predict(x_test)

print(y_pred_nb)
from sklearn.metrics import accuracy_score, classification_report, confusion_matrix

print("Naive Bayes Accuracy:", accuracy_score(y_test, y_pred_nb))

print("\nClassification Report:")
print(classification_report(y_test, y_pred_nb))

print("\nConfusion Matrix:")
print(confusion_matrix(y_test, y_pred_nb))
```

```python
results = pd.DataFrame({
    "Model": ["Decision Tree", "Naive Bayes"],
    "Accuracy": [
        accuracy_score(y_test, y_pred_dt),
        accuracy_score(y_test, y_pred_nb)
    ]
})

print(results)
```
