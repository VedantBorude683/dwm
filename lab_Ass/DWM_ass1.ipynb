```python
import pandas as pd
df=pd.read_csv("Titanic-Dataset.csv")
df.head()
```

```python
df.tail()
df.isnull().sum()
```

```python
df.dtypes
```

```python
print(" mean age: ",df["Age"].mean())
print(" median age: ",df["Age"].median())
print("mode age: ",df["Age"].mode())
```

```python
print(df["Age"].skew())
```

```python
df["Age"]=df["Age"].fillna(df["Age"].median())
```

```python
df.isnull().sum()
```

```python
df["Cabin"].describe()
```

```python
#cabin is object->conver it to category
df["Cabin"]=df["Cabin"].astype("category")
print(df["Cabin"].dtype)
```

```python
df.isnull().sum()
```

```python
df["Embarked"]=df["Embarked"].fillna(df["Embarked"].mode()[0])
```

```python
print(df.select_dtypes(include="category").columns)
```

```python
df.dtypes
```

```python
df["Survived"]=df["Survived"].astype("bool")# convert survived to bool
df["Pclass"]=df["Pclass"].astype("category")
df["Sex"]=df["Sex"].astype("category")
df["Embarked"]=df["Embarked"].astype("category")

```

```python
df.dtypes
```

```python
df.isnull().sum()
```

```python
df["Sex_Encoded"]=df["Sex"].map({"male":0,"female":1})
print(df[["Sex","Sex_Encoded"]].head())
```

```python
df=pd.get_dummies(df,columns=["Embarked"],dtype=int)
```

```python
print(df.filter(like="Embarked_").head())
```

```python
print("age range")
print("min: ",df["Age"].min())
print("max: ",df["Age"].max())


print("\nFare range:")
print("Min:", df["Fare"].min())
print("Max:", df["Fare"].max())
```

```python
from sklearn.preprocessing import MinMaxScaler
scaler=MinMaxScaler()
df[["Age_Normalized","Fare_Normalized"]]=scaler.fit_transform(df[["Age","Fare"]])
print(df[["Age","Age_Normalized","Fare","Fare_Normalized"]].head())
```
